```python
#Import lib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import yfinance as yf
```

```python
tiker = input("enter the stock symbol:")
df = yf.download(tiker,start="2025-01-01",end = "2026-01-01",interval = "1d")
print(df.head(20))
```

```python
daily_close = df["Close"]
daily_close.head(20)
```

```python
df["7_day_MA"] = df["Close"].rolling(window =7).mean()
df["21_day_MA"] = df["Close"].rolling(window =21).mean()
df.head(10)
```

```python
#Vizualisation of 7-dayMA,21_day_MA
plt.figure(figsize=(10,6))
plt.plot(df["Close"],label = "Daily close Price")
plt.plot(df["7_day_MA"],label = "7_day_MA")
plt.plot(df["21_day_MA"],label = "21_day_MA")
plt.title(f'{tiker}:Stock Price Time Serise Analysis')
plt.xlabel("Time line")
plt.ylabel("Price")
plt.grid(True)
plt.legend()
plt.show()
```
